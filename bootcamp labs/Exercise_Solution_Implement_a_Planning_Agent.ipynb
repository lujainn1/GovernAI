```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain.agents import create_react_agent, AgentExecutor

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

print("LLM ready")
```

## TODO: Create the Planning Agent

```python
# TODO 1: Create prompt for Planning Agent
planner_prompt = ChatPromptTemplate.from_template(
    """You are a Planning Agent. Break down the user's request into a clear, numbered step-by-step plan.

User Request: {input}

Create 4 to 6 logical steps. Make them actionable and specific.
Return ONLY the numbered plan."""
)

# TODO 2: Create the planner chain
planner_chain = planner_prompt | llm   # ← Your code here (prompt | llm)

print("Planning Agent setup - complete the TODOs")


```

## TODO: Create a Simple Execution Agent

```python
# Simple Execution Agent (simulated - no real tools needed for this exercise)
executor_prompt = ChatPromptTemplate.from_template(
    """You are an Execution Agent.
Here is the plan from the Planner:

{plan}

Simulate executing each step and give a short status update for every step."""

)

# TODO 3: Create executor chain
executor_chain = executor_prompt | llm   # ← Your code here (prompt | llm)

print("Execution Agent setup - complete the TODOs")



```

## TODO: Implement Plan Generation and Delegation

```python
def run_planning_exercise(user_query: str):
    print("Starting Planning Agent Exercise\n")

    plan_result = planner_chain.invoke({"input": user_query})
    plan = plan_result.content if hasattr(plan_result, "content") else str(plan_result)

    print("Generated Plan:")
    print(plan)
    print("\n" + "-"*60 + "\n")

    execution_result = executor_chain.invoke({"plan": plan})

    print("Execution Agent Output:")
    print(execution_result.content if hasattr(execution_result, "content") else execution_result)

    print("\n" + "="*70)
    print("Exercise Completed!")

# Test
user_request = "Create a marketing campaign plan for launching a new AI writing assistant tool."
run_planning_exercise(user_request)
```
